```python
import pandas as pd
df = pd.read_csv("https://github.com/datanizing/azk-ki/raw/refs/heads/main/meteo_api.csv.gz")
df
```

```python
df["Location"].value_counts()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
dfc = df[df["Location"] == "Canberra"]
```

```python
dfct = dfc.set_index("Date")[["temperature_2m_max", "temperature_2m_min"]]
dfct.plot()
```

```python
from prophet import Prophet
```

```python
df = pd.DataFrame({ "ds": dfct.index, "y": dfct["temperature_2m_max"]})
df
```

```python
m = Prophet()
m.fit(df)
```

```python
future = m.make_future_dataframe(periods=730)
future.tail()
```

```python
forecast = m.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```

```python
fig1 = m.plot(forecast)
```

```python
fig2 = m.plot_components(forecast)
```
